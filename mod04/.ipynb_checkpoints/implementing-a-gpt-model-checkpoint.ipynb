## 4.1 Building the GPT Architecture

### 4.1.1 What we are building?

Imagine we want to train a model to continue this Nigerian sentence:

> "Abeg, I don send the money since…"

Our GPT model needs to transform the input tokens into contextual representations and then produce scores for possible next tokens.

The architecture follows this flow:

<img src="./illustrations/4_1_1-what-are-we-building.png" width=700px>

For this first lesson, we'll implement the architecture's overall data flow using placeholder Transformer components. In later lessons, we'll replace the placeholders with the actual components we've studied.

### 4.1.2 Set up the configuration

A GPT model has several configurable dimensions. We'll use a small configuration for learning.

````python
GPT_CONFIG = {
    "vocab_size": 1000,
    "context_length": 128,
    "emb_dim": 64,
    "n_heads": 4,
    "n_layers": 2,
    "drop_rate": 0.1,
    "qkv_bias": False,
}
````

These values are deliberately small so the example is easy to understand. They are not the configuration of a production GPT model.

<img src="./illustrations/4_1_2-set-up-the-configuration.png" width=700px>

### 4.1.3 Create placeholder components

We need the complete architecture before implementing every internal component.

For now, the Transformer block and final normalization will simply pass their inputs through unchanged.

````python
import torch
import torch.nn as nn


class DummyTransformerBlock(nn.Module):
    def __init__(self, cfg):
        super().__init__()

    def forward(self, x):
        return x


class DummyLayerNorm(nn.Module):
    def __init__(self):
        super().__init__()

    def forward(self, x):
        return x
````

#### Why use placeholders?

They allow us to test the architecture's data flow without getting distracted by the internal calculations. They do not perform attention, normalization or feed-forward transformations yet.

### 4.1.4 Implement the GPT model

Now we connect the embedding layers, Transformer blocks, normalization and output projection.

````python
class DummyGPTModel(nn.Module):
    def __init__(self, cfg):
        super().__init__()

        self.tok_emb = nn.Embedding(
            cfg["vocab_size"],
            cfg["emb_dim"]
        )

        self.pos_emb = nn.Embedding(
            cfg["context_length"],
            cfg["emb_dim"]
        )

        self.drop_emb = nn.Dropout(
            cfg["drop_rate"]
        )

        self.trf_blocks = nn.Sequential(
            *[
                DummyTransformerBlock(cfg)
                for _ in range(cfg["n_layers"])
            ]
        )

        self.final_norm = DummyLayerNorm()

        self.out_head = nn.Linear(
            cfg["emb_dim"],
            cfg["vocab_size"],
            bias=False
        )

    def forward(self, in_idx):
        batch_size, seq_len = in_idx.shape

        tok_embeds = self.tok_emb(in_idx)

        pos_embeds = self.pos_emb(
            torch.arange(
                seq_len,
                device=in_idx.device
            )
        )

        x = tok_embeds + pos_embeds
        x = self.drop_emb(x)
        x = self.trf_blocks(x)
        x = self.final_norm(x)

        logits = self.out_head(x)

        return logits
````

Let's understand the important parts.
- self.tok_emb converts token IDs into vectors.
- self.pos_emb provides a learned vector for each position.
- self.drop_emb applies dropout during training.
- self.trf_blocks stacks the configured number of Transformer blocks.
- self.final_norm normalizes the final representations.
- self.out_head projects each representation into a vector with `vocab_size` scores.
  
Notice that the output projection maps from emb_dim to vocab_size. It doesn't directly predict a word; it produces **logits**, which can later be converted into probabilities.

### 4.1.5 Test the architecture

Let's feed the model a batch of token IDs.

````python
torch.manual_seed(123)

batch_size = 2
seq_len = 5

# Generate a tensor filled with random integers drawn uniformly between a 
# lower bound (inclusive) and an upper bound (exclusive)
batch = torch.randint(
    0,                        # Lower bound
    GPT_CONFIG["vocab_size"], # Upper bound
    (batch_size, seq_len)     # Size - Define the shape of the tensor
)

model = DummyGPTModel(GPT_CONFIG)

logits = model(batch)

print("Input shape:", batch.shape)
print("Output shape:", logits.shape)
````

### 4.1.6 Why is the output three-dimensional?

The output shape is:

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $ (B,T,V) $

where:
- $ B=2 $: two sequences in the batch.
- $ T=5 $: five tokens per sequence.
- $ V=1000 $: one logit for every token in the illustrative vocabulary.
  
So the model produces 1,000 scores for each of the five input positions in each sequence.

During next-token generation, we usually take the logits at the final position to choose the next token. During training, logits at all positions can be used to calculate the next-token prediction loss.

One important detail: because our Transformer blocks are placeholders, this test confirms that the architecture and tensor dimensions work. It does not mean we've built a functional text-generating GPT yet.

## 4.2 Normalizing Activations with Layer Normalization

In our previous section, we assembled the overall GPT architecture using placeholder components.

One of those placeholders was:

````python
class DummyLayerNorm(nn.Module):
    def __init__(self):
        super().__init__()

    def forward(self, x):
        # Pass the input through unchanged for now
        return x
````

Now, we'll replace it with an actual Layer Normalization implementation.

### 4.2.1 Why do we need Layer Normalization?

Imagine GPT receives this sentence:

> "Abeg, I don send the money since morning."

After tokenization and embedding, each token is represented by a vector of numerical values.

For example, a simplified token representation might look like this:

````python
# An illustrative feature vector for one token
x = torch.tensor([
    0.8, 1.2, -0.4, 2.1, 0.5
])
print(x)
````

These values represent learned features. They are not human-readable meanings.

As information flows through the Transformer, the numerical activations can develop different scales and distributions.

Layer Normalization helps stabilize these activations by normalizing the feature values within each token representation.

This makes optimisation more stable during training.

### 4.2.2 How does Layer Normalization work?

Layer Normalization involves four main operations:

1. Calculate the mean.
2. Calculate the variance.
3. Normalize the values.
4. Apply learnable scale and shift parameters.

Let's examine each operation.

#### Step 1: Calculate the mean

The mean is the average of the values in the feature vector.

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $ \mu = \frac{1}{d}\sum\limits_{i=1}^{d}x_i $

````python
import torch

# Create an illustrative token representation
x = torch.tensor([0.8, 1.2, -0.4, 2.1, 0.5])

# Calculate the average of the feature values
mean = x.mean()

print("Mean:", mean)
````

#### Step 2: Calculate the variance

Variance measures how far the feature values spread around their mean.

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $ \sigma^2 = \frac{1}{d} \sum\limits_{i=1}^d(x_i - \mu)^2 $

````python
# Calculate the variance around the mean
variance = x.var(unbiased=False)

print("Variance:", variance)
````

We use `unbiased=False` because Layer Normalization uses the population variance over the feature dimensions, rather than the sample variance. This is referring directly to the denominator used when calculating variance ($ N $ vs. $ N - 1$).

**Why LayerNorm uses population variance**: LayerNorm views the $ N $ features of a single data point (for example, all 768 numbers in a Transformer token embedding) as the **entire set** of values being normalized. Because you have every single feature right there in memory—not a small subset drawn from a larger group—you divide directly by $ N $ (the total number of features).

#### Step 3: Normalize the values

We subtract the mean from every feature and divide by the standard deviation.

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $ \hat{x}_i = \frac{x_i - \mu}{\sqrt{\sigma^2 + \epsilon}} $

Here, $ \epsilon $ is a small numerical constant that prevents division by zero.

````python
# A small constant for numerical stability
eps = 1e-5

# Normalize the feature values
x_norm = (x - mean) / torch.sqrt(variance + eps)

print("Normalized values:", x_norm)
print("Normalized mean:", x_norm.mean())
print("Normalized variance:", x_norm.var(unbiased=False))
````

After normalization, the feature values should have a mean close to zero and a variance close to one.

#### Step 4: Apply learnable scale and shift parameters

Normalization alone is not the complete operation.

Layer Normalization introduces two learnable parameters:
- Gamma ($\gamma$): controls the scale.
- Beta ($\beta$): controls the shift.

The final operation is:

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $ y = \gamma\hat{x} + \beta $

````python
# Initialize learnable scale and shift parameters
gamma = torch.ones_like(x)
beta = torch.zeros_like(x)

# Apply the learnable transformation
y = gamma * x_norm + beta
print("Final output:", y)
````

Initially, gamma is 1 and beta is 0, so the output is unchanged from the normalized values.

During training, these parameters are updated through backpropagation.

### 4.2.3 Implement Layer Normalization as a PyTorch module

We've implemented the mathematical operations individually. Now let's combine them into a reusable PyTorch module.

````python
import torch
import torch.nn as nn


class LayerNorm(nn.Module):
    def __init__(self, emb_dim):
        super().__init__()

        # Learnable scale parameter, initialised to 1
        self.eps = 1e-5
        self.scale = nn.Parameter(torch.ones(emb_dim))

        # Learnable shift parameter, initialized to 0
        self.shift = nn.Parameter(torch.zeros(emb_dim))

    def forward(self, x):
        # Calculate the mean across the feature dimension
        mean = x.mean(dim=-1, keepdim=True)

        # Calculate population variance across the feature dimension
        var = x.var(dim=-1, keepdim=True, unbiased=False)

        # Normalize the input features
        norm_x = (x - mean) / torch.sqrt(var + self.eps)

        # Apply learnable scale and shift
        return self.scale * norm_x + self.shift
````

Let's understand two important details.

**Why** `dim=-1`**?**

Our GPT input has the shape:

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $ (B, T, D) $

- $ B $: batch size
- $ T $: sequence length
- $ D $: embedding dimension

`dim=-1` tells PyTorch to calculate the mean and variance across the embedding dimension.

This means each token is normalized independently.

**Why** `keepdim=True`**?**

It preserves the dimension of the calculated mean and variance, allowing PyTorch to broadcast them correctly during subtraction and division.

### 4.2.4 Test our Layer Normalization

Let's create a batch containing two Nigerian sentences, represented here by random token features.

````python
# Set a seed so the example is reproducible
torch.manual_seed(123)

# Create a batch of 2 sequences, each containing 5 tokens
# Each token has 64 embedding features
batch = torch.randn(2, 5, 64)

# Create our Layer Normalization module
layer_norm = LayerNorm(64)

# Pass the batch through Layer Normalization
output = layer_norm(batch)

# Display the input and output dimensions
print("Input shape:", batch.shape)
print("Output shape:", output.shape)

# Examine the normalized mean and variance of one token
print("First token mean:", output[0, 0].mean().item())
print("First token variance:", output[0, 0].var(unbiased=False).item())
````

The output retains exactly the same dimensions as the input.

### 4.2.5 Replace the placeholder in our GPT architecture

Now we can replace `DummyLayerNorm` with our actual implementation.

In our previous `DummyGPTModel`, change:

````python
class DummyGPTModel(nn.Module):
    def __init__(self, cfg):
        super().__init__()

        self.tok_emb = nn.Embedding(
            cfg["vocab_size"],
            cfg["emb_dim"]
        )

        self.pos_emb = nn.Embedding(
            cfg["context_length"],
            cfg["emb_dim"]
        )

        self.drop_emb = nn.Dropout(
            cfg["drop_rate"]
        )

        self.trf_blocks = nn.Sequential(
            *[
                DummyTransformerBlock(cfg)
                for _ in range(cfg["n_layers"])
            ]
        )

        # NEW - Create a real Layer Normalization layer
        self.final_norm = LayerNorm(cfg["emb_dim"])

        self.out_head = nn.Linear(
            cfg["emb_dim"],
            cfg["vocab_size"],
            bias=False
        )

    def forward(self, in_idx):
        batch_size, seq_len = in_idx.shape

        tok_embeds = self.tok_emb(in_idx)

        pos_embeds = self.pos_emb(
            torch.arange(
                seq_len,
                device=in_idx.device
            )
        )

        x = tok_embeds + pos_embeds
        x = self.drop_emb(x)
        x = self.trf_blocks(x)
        x = self.final_norm(x)

        logits = self.out_head(x)

        return logits
````

The rest of the GPT architecture remains unchanged for now.

#### What we've implemented

**Layer Normalization**

- Mean calculation
- Variance calculation
- Feature normalization
- Learnable scale and shift
- PyTorch implementation
- Integration into the GPT architecture

#### Key takeaway

Layer Normalization operates independently on each token's feature representation. It does not normalize across all tokens in a sequence.

For example, in:

> "Abeg, I don send the money since morning."

Each token has its own embedding representation, and Layer Normalization normalizes the features within that representation.

## 4.3 Implementing a Feed-Forward Network with GELU Activations

In our previous section, we implemented Layer Normalization and integrated it into our GPT architecture.

> Now, we'll implement another important Transformer component: the **Feed-Forward Network (FFN)**.

Remember our example:
"The trader reduced the price because the customer bought plenty of tomatoes."

The attention mechanism helps each token gather contextual information from other tokens. The feed-forward network then transforms the representation at each token position independently.

Let's implement it.

### 4.3.1 Understanding the architecture

The feed-forward network used in the GPT architecture consists of:
1. A first linear layer that expands the embedding dimension.
2. A GELU activation function.
3. A second linear layer that projects the expanded representation back to the original embedding dimension.

<img src="./illustrations/4_3_1-understanding-the-architecture.png" width=700px>

For our example, we'll use an embedding dimension of 64 and expand it to 256.

The expansion factor is four:

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $ D_\text{hidden} = 4D $

### 4.3.2 Implement the Feed-Forward Network

Let's start with the PyTorch implementation.

````python
import torch
import torch.nn as nn

class FeedForward(nn.Module):
    def __init__(self, cfg):
        super().__init__()

        # Get the embedding dimension fro the GPT configuration
        self.layers = nn.Sequential(

            # First linear layer: expand the embedding dimension
            nn.Linear(
                cfg["emb_dim"],
                4 * cfg["emb_dim"]
            ),

            # Apply GELU activation to introduce non-linearity
            nn.GELU(),

            # Second linear layer: project back to the original dimension
            nn.Linear(
                4 * cfg["emb_dim"],
                cfg["emb_dim"]
            ),
        )

    def forward(self, x):
        # Pass the input through both linear and GELU
        return self.layers(x)
````

Let's break down the implementation.

**First linear layer**

```python
nn.Linear(
    cfg["emb_dim"],
    4 * cfg["emb_dim"]
)
```

If our embedding dimension is 64, this layer transforms each token representation from 64 features to 256 features.

Mathematically:

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $ h = xW_1 + b_1 $

This gives the network a larger intermediate feature space in which to transform the representation.

**GELU activation**

```python
nn.GELU()
```

GELU introduces non-linearity into the network.

Without an activation function between the two linear layers, combining them would essentially produce a single linear transformation.

GELU is defined as:

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $ GELU(x) = x\Phi(x) $

Where $ \Phi(x) $ is the cumulative distribution function of the standard normal distribution.

Unlike ReLU, GELU does not simply discard every negative input. It smoothly scales inputs according to their values.

**Second linear layer**

```python
nn.Linear(
    4 * cfg["emb_dim"],
    cfg["emb_dim"]
)
```

This projects the expanded 256-dimensional representation back to 64 dimensions.

The output therefore has the same embedding dimension as the input.

The complete mathematical operation is:

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $ FFN(x) = W_2GELU(W_1x + b_1) + b_2 $

### 4.3.3 Test the Feed-Forward Network

Let's test our implementation using the same tensor dimensions we used in the Layer Normalization lesson.

````python
# Set a random seed to make the example reproducible
torch.manual_seed(123)

# Create a batch of 2 sequences, eac containing 5 tokens
# Each token has 64 embedding features
batch = torch.randn(2, 5, 64)

# Create the Feed-Forward Network using our GPT configuration
ffn = FeedForward(GPT_CONFIG)

# Pass the input through the Feed-Forward Network
output = ffn(batch)

# Display the input and output dimensions
print("Input shape:", batch.shape)
print("Output shape:", output.shape)
````

Notice that the input and output shapes are identical.

Although the intermediate representation expands to 256 dimensions, the second linear layer returns it to 64 dimensions.

### 4.3.4 Examine the intermediate dimensions

Let's make the dimensional transformation more explicit.

````python
# Extract the first linear layer from the network
first_layer = ffn.layers[0]

# Extract the GELU activation
gelu_layer = ffn.layers[1]

# Extract the second linear layer
second_layer = ffn.layers[2]

# Apply the first linear transformation
expanded = first_layer(batch)

# Apply GELU activation
activated = gelu_layer(expanded)

# Apply the second linear transformation
projected = second_layer(activated)

# Display the shape after each operation
print("Original input:", batch.shape)
print("After first linear layer:", expanded.shape)
print("After GELU:", activated.shape)
print("After second linear layer:", projected.shape)
````

GELU does not change the dimensions. It transforms the numerical values.

### 4.3.5 Replace the placeholder in our GPT architecture

In our previous architecture, we created `DummyTransformerBlock` as a placeholder.

We haven't implemented the actual Transformer block yet, so we won't replace that component now.

However, our FeedForward class is now ready to be used inside it.

When we implement the real Transformer block, we'll connect:
- Layer Normalization
- Multi-head attention
- Residual connections
- Feed-Forward Network

The feed-forward network will operate on each token's representation independently. It doesn't mix information between different token positions; that is the attention mechanism's job.

**Feed-Forward Network implementation**

We have:
- Implemented the first linear layer.
- Expanded the embedding dimension fourfold.
- Applied GELU activation.
- Implemented the second linear layer.
- Projected back to the original embedding dimension.
- Tested the network using PyTorch tensors.

## 4.4 Implementing Shortcut Connections (Residual Connections)

In our previous section, we implemented the Feed-Forward Network with GELU activation.

Now, we'll implement **shortcut connections**, also known as residual connections.

Remember our Nigerian example:
"Omo, fuel don cost again."

As information passes through the Transformer, each layer transforms the token representations. But what happens if these transformations make it difficult for the network to preserve useful information from earlier layers?

Shortcut connections help address this problem by allowing the original input to bypass a transformation and be added to its output.

### 4.4.1 Understand what we're implementing

A residual connection follows this simple equation:

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $ y = F(x) + x $

Where:
- $ x $ is the original input.
- $ F(x) $ is the output of a transformation, such as attention or a feed-forward network.
- $ y $ is the resulting representation.

<img src="./illustrations/4_4_1-understanding-what-we-are-implementing.png" width=700px>

The original input has a direct path to the output, bypassing the transformation.

### 4.4.2 Implement a basic residual connection

Let's begin with a simple example.

We'll create a transformation using a linear layer and then add the original input to its output.

````python
import torch
import torch.nn as nn

# Define a simple transformation with a linear layer
class ExampleTransformation(nn.Module):
    def __init__(self, emb_dim):
        super().__init__()

        # Create a linear layer that preserves the input dimension
        self.linear = nn.Linear(emb_dim, emb_dim)

    def forward(self, x):
        # Transform the input using the linear layer
        return self.linear(x)

        
# Define a residual connection
class ResidualConnection(nn.Module):
    def __init__(self, layer):
        super().__init__()

        # Store the transformation that will be wrapped
        self.layer = layer

    def forward(self, x):
        # Calculate the transformed representation
        transformed = self.layer(x)

        # Add the original input to the transformed representation
        return x + transformed
````

Let's understand what is happening.

````python
# Create a transformation for 64-dimensional token representation
transformation = ExampleTransformation(emb_dim=64)

# Wrap the transformation with a residual connection
residual_layer = ResidualConnection(transformation)

# Create a batch of 2 sequences, each containing 5 tokens
# Each token has 64 features
batch = torch.randn(2, 5, 64)

# Pass the input through the residual connection
output = residual_layer(batch)

# Display the input and output dimensions
print("Input shape:", batch.shape)
print("Output shape:", output.shape)
````

Notice that the dimensions remain unchanged.

This is important because the original input and transformed output must have compatible dimensions for element-wise addition.

### 4.4.3 Why do we need dropout?

In our GPT configuration, we defined:

```python
# Dropout probability used in our GPT configuration
GPT_CONFIG["drop_rate"] = 0.1
```

Dropout randomly sets some activations to zero during training. This helps reduce over-reliance on particular activations and can improve generalisation.

In the residual connection, we can apply dropout to the transformed output before adding the original input.

The operation becomes:

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $ y = x + \text{Dropout}(F(x)) $

Let's update our implementation.

````python
class ResidualConnection(nn.Module):
    def __init__(self, layer, drop_rate):
        super().__init__()

        # Store the transformation being wrapped
        self.layer = layer

        # NEW - Create a dropout layer using the configured dropout probability
        self.dropout = nn.Dropout(drop_rate)

    def forward(self, x):
        # Apply the transformation to the original input
        transformed = self.layer(x)

        # NEW - Apply dropout to the transformed output
        transformed = self.dropout(transformed)

        # Add the original input to the transformed output
        return x + transformed
````

**Important**: Dropout is active during training but automatically disabled when we call `model.eval()`.

### 4.4.4 Test the residual connection

Let's test the updated implementation.

````python
# Set a random seed for reproducibility
torch.manual_seed(123)

# Configure the embedding dimension and dropout probability
emb_dim = 64
drop_rate = 0.1

# Create the transformation
transformation = ExampleTransformation(emb_dim)

# Wrap it with our residual connection
residual_layer = ResidualConnection(
    transformation,
    drop_rate
)

# Create a batch of token representations
batch = torch.randn(2, 5, emb_dim)

# Set the module to training mode so dropout is active
residual_layer.train()

# Pass the batch through the residual connection
output = residual_layer(batch)

# Display the tensor dimensions
print("Input shape:", batch.shape)
print("Output shape:", output.shape)

# Switch to evaluation mode to disable dropout
residual_layer.eval()

# Run the same input again without dropout
output_eval = residual_layer(batch)

# Display the evaluation output dimensions
print("Evaluation output shape:", output_eval.shape)
````

The numerical outputs will differ between training and evaluation because dropout is active only during training.

### 4.4.5 How does this help with gradient flow?

One important reason for residual connections is that they provide a direct path for gradients during backpropagation.

Recall:

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $ y = F(x) + x $

Differentiating with respect to $ x $:

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $ \frac{\partial y}{\partial x} = \frac{\partial F(x)}{\partial x} + 1$

The additional $ 1 $ represents the direct contribution from the shortcut path.

This helps gradients flow through deep networks, even when the transformation itself has a small gradient.

It doesn't guarantee that vanishing gradients will never occur, but it makes optimisation of deeper networks easier.

#### What we've accomplished

**Shortcut connection implementation**

We've implemented:
- A basic residual connection.
- Addition of the original input to a transformed representation.
- Dropout on the transformed output.
- Training and evaluation behaviour.
- The relationship between residual connections and gradient flow.

## 4.5 Connecting Attention and Linear Layers in a Transformer Block

In our previous sections, we implemented:
1. Layer Normalization.
2. Feed-Forward Networks with GELU.
3. Shortcut (residual) connections.

Now, we're going to bring these components together to implement the **Transformer Block**.

Remember our example:
> "Abeg, I sent the money yesterday, but the person never acknowledge am."

The attention mechanism allows the tokens to gather relevant contextual information from other tokens. The feed-forward network then transforms the resulting representations.

The Transformer block combines both operations, using layer normalization and shortcut connections.

### 4.5.1 Understand the architecture

We'll implement the pre-layer-normalization arrangement we have studied.

<img src="./illustrations/4_5_1-connecting-attention-and-linear-layers-in-a-transformer-block.png" width=700px>

There are two separate shortcut connections:
- The first surrounds the attention mechanism.
- The second surrounds the feed-forward network.

Both preserve the original representation entering their respective transformations.

### 4.5.2 Prerequisites: Components from our previous sections and module

Before implementing the Transformer block, we need the following classes:

- `LayerNorm` — from Section 4.2.
- `FeedForward` — from Section 4.3.
- `MultiHeadAttention` — from our previous attention implementation in Module 3.
  
For this implementation, we'll use the MultiHeadAttention interface from our earlier attention lessons:

```python
MultiHeadCausalSelfAttention(
    d_in,
    d_out,
    context_length,
    dropout,
    num_heads,

```

It must implement causal multi-head self-attention and return an output with the same sequence length and embedding dimension as its input.

### 4.5.3 Implement the Transformer Block

Let's now implement the complete Transformer block.

````python
import torch
import torch.nn as nn
from multi_head_attention import MultiHeadCausalSelfAttention

class TransformerBlock(nn.Module):
    def __init__(self, cfg):
        super().__init__()

        # Create the multi-head causal self-attention mechanism
        self.att = MultiHeadCausalSelfAttention(
            d_in=cfg["emb_dim"],
            d_out=cfg["emb_dim"],
            context_length=cfg["context_length"],
            num_heads=cfg["n_heads"],
            dropout=cfg["drop_rate"],
            qkv_bias=cfg["qkv_bias"]
        )

        # Create the feed-forward network
        self.ff = FeedForward(cfg)

        # Create the first Layer Normalization layer
        self.norm1 = LayerNorm(cfg["emb_dim"])

        # Create the second Layer Normalization layer
        self.norm2 = LayerNorm(cfg["emb_dim"])

        # Apply dropout to the outputs of attention and feed-forward layers
        self.dropout = nn.Dropout(cfg["drop_rate"])

    def forward(self, x):

        # -----------------------------------------
        # First sub-block: Multi-Head Self-Attention
        # -----------------------------------------

        # Save the original input for the shortcut connection
        shortcut = x

        # Normalize the input before applying attention
        x = self.norm1(x)

        # Pass the normalized input through multi-head attention
        x = self.att(x)

        # Apply dropout to the attention output
        x = self.dropout(x)

        # Add the original input through the residual connection
        x = x + shortcut

        # -----------------------------------------
        # Second sub-block: Feed-Forward Network
        # -----------------------------------------

        # Save the output of the attention sub-block
        shortcut = x

        # Normalize the representation before the feed-forward network
        x = self.norm2(x)

        # Transform the token representations using the feed-forward network
        x = self.ff(x)

        # Apply dropout to the feed-forward output
        x = self.dropout(x)

        # Add the original representation through the residual connection
        x = x + shortcut

        # Return the transformed token representations
        return x
````

Let's examine the important parts.

#### A. Multi-head attention

```python
self.att = MultiHeadCausalSelfAttention(
    d_in=cfg["emb_dim"],
    d_out=cfg["emb_dim"],
    context_length=cfg["context_length"],
    num_heads=cfg["n_heads"],
    dropout=cfg["drop_rate"],
    qkv_bias=cfg["qkv_bias"]
)
```

We initialise multi-head attention using the configuration.

For our current configuration:
- Input dimension: 64
- Output dimension: 64
- Number of heads: 4
- Context length: 128

Each attention head processes part of the embedding representation. Their outputs are combined to produce the final attention output.

The causal mask ensures that a token cannot attend to future tokens.

#### B. First shortcut connection

```python
shortcut = x

# Normalize the input
x = self.norm1(x)

# Apply attention
x = self.att(x)

# Apply dropout
x = self.drop_shortcut(x)

# Add the original input
x = x + shortcut
```

Notice that we save the original input before applying normalization and attention.

After the attention operation, we add the saved representation to the attention output.

Mathematically:

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $ x' = x + \text{Dropout}(\text{Attention}(\text{Norm}(x))) $

#### C. Second shortcut connection

This follows the same principle, but now the transformation is the feed-forward network.

Mathematically:

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $ y = x' + \text{Dropout}(\text{FFN}(\text{Norm}(x'))) $

The output of the first sub-block becomes the input of the second.

The input and output dimensions are identical.

This is important because the Transformer block must preserve the embedding dimension so that multiple blocks can be stacked.

### 4.5.5 Stack multiple Transformer blocks

One Transformer block performs a single round of attention-based contextual processing and feed-forward transformation.

GPT uses multiple Transformer blocks stacked together.

Remember our configuration:

```python
# Number of Transformer blocks in our GPT model
GPT_CONFIG["n_layers"] = 2
```

We can create two blocks using `nn.Sequential`.

````python
# Create a sequence containing the configured number of Transformer blocks
transformer_blocks = nn.Sequential(
    *[
        TransformerBlock(GPT_CONFIG)
        for _ in range(GPT_CONFIG["n_layers"])
    ]
)

# Pass the input through both Transformer blocks
output = transformer_blocks(batch)

# Display the resulting dimensions
print("Input shape:", batch.shape)
print("Output shape:", output.shape)
````

Even though the input passes through two Transformer blocks, its dimensions remain unchanged.

However, the numerical representations are progressively transformed by the attention and feed-forward operations in each block.

### 4.5.6 What happens to our Nigerian sentence?

Consider:
"Abeg, I sent the money yesterday, but the person never acknowledge am."

Let's follow what happens conceptually:
1. The attention mechanism allows the token acknowledge to gather contextual information from earlier tokens, including money and sent.
2. The first shortcut connection preserves the representation that entered the attention sub-block.
3. The feed-forward network transforms the resulting representation.
4. The second shortcut connection preserves the representation entering the feed-forward sub-block.
   
The output is a set of contextual token representations that the next Transformer block can process.

Remember, the feed-forward network does not independently decide which other tokens to look at. That contextual interaction happens in attention.

## 4.6 Implementing the Complete GPT Model

In our previous sections, we've implemented the individual components that make up a GPT Transformer block:
1. Layer Normalization.
2. Feed-Forward Network with GELU.
3. Shortcut (residual) connections.
4. Transformer Block with multi-head causal self-attention.

We've also assembled the Transformer block by connecting these components.

Now, we'll bring everything together to implement the **complete GPT model**.

Our objective is to replace the placeholder components from Section 4.1 with the actual implementations we've built.

### 4.6.1 Review the complete GPT architecture

Let's revisit the architecture.

<img src="./illustrations/4_6_1-review-the-complete-gpt-architecture.png" width=700px>

The model receives token IDs and produces logits representing scores for every token in its vocabulary at every sequence position.

### 4.6.2 Prepare the GPT configuration

We'll continue using the small configuration from our previous sections.

````python
# Define the configuration of our GPT model
GPT_CONFIG = {
    "vocab_size": 1000,       # Number of tokens in the vocabulary
    "context_length": 128,    # Maximum number of tokens the model can process
    "emb_dim": 64,            # Dimensionality of each token embedding
    "n_heads": 4,             # Number of attention heads
    "n_layers": 2,            # Number of Transformer blocks
    "drop_rate": 0.1,         # Dropout probability
    "qkv_bias": False         # Whether Q, K, and V projections use bias
}
````

These are illustrative dimensions for learning, not the dimensions of a production GPT model.

### 4.6.3 Implement the complete GPT model

We'll now create a `GPTModel` class.

Unlike our earlier `DummyGPTModel`, this implementation uses the actual `TransformerBlock` and `LayerNorm` classes we developed.

````python
import torch
import torch.nn as nn


class GPTModel(nn.Module):
    def __init__(self, cfg):
        super().__init__()

        # Create the token embedding layer
        # Converts each token ID into a learnable embedding vector
        self.tok_emb = nn.Embedding(
            cfg["vocab_size"],
            cfg["emb_dim"]
        )

        # Create the positional embedding layer
        # Gives each token position its own learnable embedding
        self.pos_emb = nn.Embedding(
            cfg["context_length"],
            cfg["emb_dim"]
        )

        # Apply dropout to the combined token and positional embeddings
        self.drop_emb = nn.Dropout(
            cfg["drop_rate"]
        )

        # Stack the configured number of actual Transformer blocks
        self.trf_blocks = nn.Sequential(
           *[
               TransformerBlock(cfg)
               for _ in range(cfg["n_layers"])
           ] 
        )

        # Normalize the final Transformer representations
        self.final_norm = LayerNorm(
            cfg["emb_dim"]
        )

        # Project the final representations into vocabulary-sized logits
        self.out_head = nn.Linear(
            cfg["emb_dim"],
            cfg["vocab_size"],
            bias=False
        )

    def forward(self, in_idx):

        # Get the batch size and sequence length from the input token IDs
        batch_size, seq_len = in_idx.shape

        # Convert token IDs into token embeddings
        tok_embeds = self.tok_emb(in_idx)

        # Generate position indices for the input sequence
        positions = torch.arange(
            seq_len,
            device=in_idx.device
        )

        # Convert position indices into positional embeddings
        pos_embeds = self.pos_emb(positions)

        # Combine token embeddings and positional embeddings
        x = tok_embeds + pos_embeds

        # Apply dropout to the combined embeddings
        x = self.drop_emb(x)

        # Pass the representations through the stacked Transformer blocks
        x = self.trf_blocks(x)

        # Normalize the final Transformer representations
        x = self.final_norm(x)

        # Convert the representations into vocabulary-sized logits
        logits = self.out_head(x)

        # Return the logits for every token position
        return logits
````

Let's examine the important sections.

#### A. Token embeddings

```python
# Convert token IDs into learnable embedding vectors
self.tok_emb = nn.Embedding(
    cfg["vocab_size"],
    cfg["emb_dim"]
)
```

Suppose our tokenizer produces these illustrative token IDs:

```python
token_ids = torch.tensor([[12, 45, 78, 91, 23]])
```

The embedding layer looks up the corresponding vector for each token ID.

With an embedding dimension of 64, each token receives a 64-dimensional representation.

#### B. Positional embeddings

```python
# Generate positional embeddings for the input sequence
positions = torch.arange(
    seq_len,
    device=in_idx.device
)

pos_embeds = self.pos_emb(positions)
```

The positional embedding layer provides a learned vector for each position.

For a sequence containing five tokens, the positions are:

```text
[0, 1, 2, 3, 4]
```

These positional embeddings are added to the token embeddings.

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $ X = E_\text{token} + E_\text{position} $

This gives the initial representation information about both token identity and position.

#### C. Stacked Transformer blocks

```python
# Process the representations through all Transformer blocks
x = self.trf_blocks(x)
```

This is where the components we implemented in Section 4.5 come into play.

Each Transformer block:
1. Normalizes the input.
2. Applies multi-head causal self-attention.
3. Adds a residual connection.
4. Normalizes the resulting representation.
5. Applies the feed-forward network.
6. Adds another residual connection.

The output of one block becomes the input of the next.

#### D. Final Layer Normalization

```python
# Normalize the output of the final Transformer block
x = self.final_norm(x)
```

After all Transformer blocks have processed the sequence, we apply another Layer Normalization.

This is the final normalization stage before the output projection.

#### E. Output projection

```python
# Convert each token representation into vocabulary-sized logits
self.out_head = nn.Linear(
    cfg["emb_dim"],
    cfg["vocab_size"],
    bias=False
)
```

This is an important part of GPT.

Our Transformer blocks produce contextual representations with 64 features per token.

However, our vocabulary contains 1,000 possible tokens.

The output projection maps each 64-dimensional representation into a vector containing 1,000 scores.

These scores are called **logits**.

They are not probabilities yet.

### 4.6.4 Test the complete GPT model

Now let's create an instance of our model and pass some token IDs through it.

````python
# Set a random seed to make the example reproducible
torch.manual_seed(123)

# Create a batch containing two sequences of five token IDs each
# These are illustrative tokeni IDs, not actual tokenizer output
batch = torch.tensor([
    [12, 45, 78, 91, 23],
    [34, 56, 89, 11, 67]
])

# Initialise the complete GPT model using our configuration
model = GPTModel(GPT_CONFIG)

# Switch the model to evaluation mode to disable dropout
model.eval()

# Disable gradient calculation because we are only testing inference
with torch.no_grad():

    # Pass the input token IDs through the GPT model
    logits = model(batch)

# Display the input and output dimensions
print("Input shape:", batch.shape)
print("Output shape:", logits.shape)
````

Let's understand this output.

<img src="./illustrations/4_6_4-test-the-complete-gpt-model.png" width=700px>

The output shape is:

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $ (B, T, V) $

This means the model produces 1,000 logits for every token position in each sequence.

### 4.6.5 Examine the logits for the final token position

Remember our objective: GPT predicts the next token based on the preceding context.

During autoregressive generation, we use the logits at the final input position to determine the next token.

Let's extract them.

````python
# Select the logits corresponding to the final token position
# The result contains one vocabulary-sized vector per sequence
next_token_logits = logits[:, -1, :]

# Display the shape of the extracted logits
print("Next-token logits shape:", next_token_logits.shape)
````

Why?
- `:` selects both sequences in the batch.
- `-1` selects the final position in each sequence.
- `:` selects all 1,000 vocabulary logits.

We now have one vector of 1,000 scores for each input sequence.

### 4.6.6 Convert logits into probabilities

Logits are raw scores. To interpret them as probabilities, we apply the softmax function.

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $ P(x_i) = \frac{e^{z_i}}{\sum_je^{z_j}} $

Where $ z_i $ is the logit for a particular vocabulary token.

````python
# Apply softmax to convert logits into probabilities
# dim=-1 applies softmax across the vocabulary dimension
probs = torch.softmax(
    next_token_logits,
    dim=-1
)

# Display the probability tensor dimensions
print("Probability shape:", probs.shape)

# Check that the probabilities for each sequence sum to approximately 1
print("Probability sums:", probs.sum(dim=-1))
````

Each sequence now has a probability distribution across the vocabulary.

**Important**: We have not trained this GPT model. Its parameters are randomly initialised, so its predictions are not meaningful yet.

### 4.6.7 Select the most likely next token

````python
# Find the vocabulary token with the highest probability
# Each result represents the predicted next token ID for one sequence
next_token_ids = torch.argmax(
    probs,
    dim=-1,
    keepdim=True
)

# Display the predicted token IDs
print("Predicted next token IDs:\n", next_token_ids)
````

Because our model is randomly initialised, these token IDs are not meaningful language predictions.

Also, `argmax` is just one possible selection strategy. Later, we'll explore sampling and other generation strategies.
